# Feature Engineering

```python

num_cols = df.select_dtypes(include= 'number')
num_cols.skew()
# near to 0 => centralized
# negative => negative
# positiive(greater than  0) => Right skewed
```

Feature Engineering

```python
top_countries = df['Country'].value_counts().index[:10].tolist()
```

```python
def group_countries(country):
  if country in top_countries:
    return country
  else:
    return 'Other'
```

```python
df['Grouped_country'] = df['Country'].apply(group_countries)
df['Grouped_country'].value_counts()
```

```python
df.columns
```

```python
skewed_cols = ['Study_Hours']
others_numeric_cols = ['Age','Avg_Daily_Usage_Hours', 'Daily_Unlocks','Physical_Activity_Hours', 'Sleep_Hours_Per_Night']
ordinal_col = ['Stress_Level']
normal_cols = ['Gender', 'Country', 'Academic_Level', 'Most_Used_Platform','Purpose_Of_Use','Grouped_country']
feature_cols = skewed_cols + others_numeric_cols + ordinal_col + normal_cols
x = df[feature_cols]
y = df['Mental_Health_Score']
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

# Creating pipeline For model

```python
# creating pipeline for better usability of code
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer , StandardScaler , OrdinalEncoder , OneHotEncoder
from sklearn.compose import ColumnTransformer
skewed_pipeline  = Pipeline(steps = [
    ('log_transform' , FunctionTransformer(np.log1p)),
    ('scaler', StandardScaler())
])

plain_pipeline  = Pipeline(steps = [
    ('scaler', StandardScaler())
])

ordinal_pipeline = Pipeline(steps = [
    ('ordinal_pipeline' , OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])

normal_pipeline = Pipeline(steps = [
  ('normal_pipeline' , OneHotEncoder(handle_unknown= 'ignore'))
])
# decide which feature goes to which column and pipeline
preprocessor = ColumnTransformer([
    ('skewed_pipeline', skewed_pipeline, skewed_cols),
    ('plain_pipeline', plain_pipeline, others_numeric_cols),
    ('ordinal_pipeline', ordinal_pipeline, ordinal_col),
    ('normal_pipeline', normal_pipeline, normal_cols)
])
```

```python
preprocessor
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score , mean_squared_error , mean_absolute_error
lr_pipeline = Pipeline(steps = [
  ('preprocessor' , preprocessor),
  ('regressor', LinearRegression())
])
lr_pipeline.fit(x_train,y_train)
lr_predict = lr_pipeline.predict(x_test)
lr_predict_train = lr_pipeline.predict(x_train)

lr_r2_testing = r2_score(y_test,lr_predict)
lr_r2_trainig = r2_score(y_train,lr_predict_train)
lr_mae = mean_absolute_error(y_test,lr_predict)

print(f"R2 Score of Testing Data : {lr_r2_testing}")
print(f"R2 Score of Training Data : {lr_r2_trainig}")
print(f"Mean Absolute Error : {lr_mae}")
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_pipeline = Pipeline(steps = [
    ('preprocessor',preprocessor) ,
    ('random_forest' , RandomForestRegressor(random_state = 42))
])
rf_pipeline.fit(x_train,y_train)
rf_predict_testing = rf_pipeline.predict(x_test)
rf_predict_training = rf_pipeline.predict(x_train)
rf_r2_testing = r2_score(y_test,rf_predict_testing)
rf_r2_training = r2_score(y_train,rf_predict_training)
rf_mae = mean_absolute_error(y_test,rf_predict_testing)
print(f"R2 Score of Testing Data : {rf_r2_testing}")
print(f"R2 Score of Training Data : {rf_r2_training}")
print(f"Mean Absolute Error : {rf_mae}")
```

Based on 2 model Random forest gives most accuracy percentage on testing data compare to linear regression model


```python
# Hyperparameter using to check the better accuracy of model on random forest
from sklearn.model_selection import RandomizedSearchCV
param_grid = {
    'random_forest__n_estimators' : [100,200,300],
    'random_forest__max_depth' :   [5,10,15] ,
    'random_forest__min_samples_split' : [2,5,10] ,
    'random_forest__min_samples_leaf' : [1,2,4]
}
random_search = RandomizedSearchCV(
    estimator = rf_pipeline ,
    param_distributions= param_grid ,
    n_iter = 5,
    cv = 5 ,
    scoring = 'r2',
    random_state = 42 ,
    n_jobs = -1
)
random_search.fit(x_train,y_train)


```

```python
random_search.best_params_
```

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_predict = rf_best_pipeline.predict(x_test)

print(f"R2 score after hypertunning : {r2_score(y_test,rf_best_predict)}")
print(f"Mean Absolute Error after hypertuning : {mean_absolute_error(y_test,rf_best_predict)}")
```

# Model Evaluation :

1.   r2_score

```python
from sklearn.metrics import mean_squared_error
lr_rmse = np.sqrt(mean_squared_error(y_test, lr_predict))

rf_rmse = np.sqrt(mean_squared_error(y_test,rf_predict_testing))

rf_tuned_predict = random_search.best_estimator_.predict(x_test)
rf_tuned_rmse = np.sqrt(mean_squared_error(y_test,rf_tuned_predict))
rf_tuned_training_predict = random_search.best_estimator_.predict(x_train)
rf_tuned_training = r2_score(y_train,rf_tuned_training_predict)
rf_tuned_mae = mean_absolute_error(y_test,rf_tuned_predict)
rf_tuned_r2 = r2_score(y_test,rf_tuned_predict)

results = pd.DataFrame({
    'Model' : ['Linear Regression','Random Forest','Random Forest Tuned'],
    'R2_Score' : [lr_r2_testing,rf_r2_testing,rf_tuned_r2],
    'Mean_Absolute_Error' : [lr_mae,rf_mae,rf_tuned_mae],
    'Root_Mean_Square_Error' : [lr_rmse,rf_rmse,rf_tuned_rmse]
})
print(results)

```
